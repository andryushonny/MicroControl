Close the Optosky software before running. Without the instrument: `mc.Microscope('sim', objective=10)`.

```python
import microcontrol as mc
mc.help()
```

```python
m = mc.Microscope(objective=10)     # give the objective that is in place now
m
```

## 1. Manual mode
Emergency stop: in any cmd window `python -m microcontrol stop`.

Arrows — one frame, `E` / `Q` — focus up / down, `Shift+F` — autofocus, `1` and `2` — area corners, `F` — one more focus point, `Esc` — exit.

```python
m.manual()
```

```python
m.measure_speed()       # motor speed by the camera -> m.motor_speed
m.offset()              # my zero here (real coordinates: m.real_position)
```

## 2. Reference frames for a dataset (on an empty area of the sample)

```python
m.white_balance()
m.empty_frame()
m.dark_frame()
```

## 3. Area scan

```python
m.plan()          # scan program in G-code
```

```python
folder = m.scan(name='Sample 1', notes='')
```

## 4. Stitching, marking, export
In the viewer: `G` — two crosses in a row (the picture turns level), `S` / `B` — select squares, `N` — square numbers, `Y` — yellow hints, `E` — export, `Esc` — exit.

```python
mc.stitch(folder)
```

```python
mc.view(folder)
```

```python
mc.export(folder)
# mc.clean(folder, confirm=True)   # delete the frames and work files, keep export/
```
