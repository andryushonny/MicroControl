Close the Optosky software before running. Without the instrument: `mc.Microscope('sim', objective=10)`.

**How work is grouped:** **Project** (a piece of work, with a description) → **Sample** (one substrate on the stage: its own selections and coordinates) → **Selection** (an area on one objective; layers: map 10x > details 50x > 100x) → **Scan** (every scan of a selection is a new version) → grid, chosen squares, drawing.
Everything lives in `projects/<project>/samples/<sample>/`. Selections are kept on disk and come back when the sample is opened again.

```python
import microcontrol as mc
mc.help()
```

## 0. Project and sample
Create new ones or open existing ones (the same call does both). The menu at the top of the windows does the same: **Open**, **New project**, **New sample**, **Info** (descriptions), **Scans**, **View**.

```python
mc.projects()                     # all projects in this folder
# mc.all_samples()                # all samples of all projects
```

```python
p = mc.Project("Graphene", "flakes for transport")      # open or create
s = p.sample("Chip 1", "SiO2 285 nm, gold crosses")      # a sample of the project: open or create
s
```

```python
m = mc.Microscope(objective=10, sample=s)   # give the objective that is in place now
# without sample=...: the last opened sample; later: m.sample = p.sample("Chip 2")
m
```

## 1. Manual mode (in a window)
Emergency stop: in any cmd window `python -m microcontrol stop`.

Arrows — move one frame, `E` / `Q` — stage up / down, `F` — autofocus, `K` — focus point here, `1` and `2` — corners of a selection, `L` — light, `I` — laser, `W` — white balance, `R` — ruler, `S` twice — scan and stitch right here, `V` — look at the result, `Ctrl+Z` — undo, `H` — all keys, `Shift+Esc` — exit.

```python
m.manual()                  # a normal window; m.manual(fullscreen=True) — the whole screen
```

```python
m.measure_speed()       # motor speed by the camera -> m.motor_speed
m.offset()              # my zero here (real coordinates: m.real_position)
```

## 2. Reference frames for a dataset (on an empty area of the sample)

```python
m.white_balance()
m.empty_frame()
m.dark_frame()
```

## 3. Scan
Only the selections of the objective in place are shot; each scan is a new version in the folder of its selection.
For details inside the map: put a larger objective in place (`M` in manual mode or `m.objective = 50`), mark small selections inside the 10x one, scan again.

```python
folder = m.scan(notes='')    # or S twice in manual mode
m.sample.selections         # selections of the sample with their scans
```

```python
m.list()                    # all scans of this sample as layers; m.open(N) — look again, m.remove(N) — delete
```

## 4. Stitching, marking, export
In the viewer: `S` — stitch here, `C` — two crosses in a row (the picture turns level), `M` / `B` — select squares, `N` — square numbers, `Y` — yellow hints, `R` — ruler, `E` — export (square pictures + drawing for AutoCAD), `Shift+Esc` — exit. The menu on top: **Open** another sample, **Scans** (every version), **Info**.

```python
mc.stitch(folder)           # or S in the viewer
```

```python
mc.view(folder)             # or: m.sample.selections[0].view()
```

```python
mc.export(folder)           # export/drawing.dxf (+ .dwg with the ODA File Converter) and export/squares/*.png
# mc.clean(folder, confirm=True)   # delete the frames and work files, keep export/
```
